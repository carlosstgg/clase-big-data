### Actividad 3 — Estadística descriptiva

```python
df_2023 = df[df["Year"] == 2023]
cambio_anual = df_2023["Annual change in population"]
```

```python
cambio_anual.mean()
```

```python
cambio_anual.median()
```

```python
cambio_anual.var()
```

```python
cambio_anual.std()
```

```python
cambio_anual.min()
```

```python
cambio_anual.max()
```

#### Preguntas
1. ¿Cuál es el valor de la media?
**Respuesta:** El valor devuelto por `cambio_anual.mean()`.

2. ¿Cuál es el valor de la mediana?
**Respuesta:** El valor devuelto por `cambio_anual.median()`.

3. ¿Cuál es el valor de la desviación estándar?
**Respuesta:** El valor devuelto por `cambio_anual.std()`.

4. ¿Qué entidad presenta el valor mínimo y cuál el máximo?
**Respuesta:** Se obtiene al buscar las entidades correspondientes a esos valores.

```python
entidad_min = df_2023[df_2023["Annual change in population"] == cambio_anual.min()]["Entity"].values[0]
entidad_max = df_2023[df_2023["Annual change in population"] == cambio_anual.max()]["Entity"].values[0]
print(entidad_min)
print(entidad_max)
```
