### Países que inician con G y estadísticas de Germany

Se identifican las entidades cuyo nombre comienza con la letra **G** (Gallegos) y se calculan las estadísticas descriptivas del cambio anual de población de **Germany**.

```python
# Países o entidades que inician con la letra G
paises_g = df.loc[df["Entity"].str.startswith("G", na=False), "Entity"].unique()
print("Países o entidades que inician con G:")
print(paises_g)
```

```python
# Cambio anual de población histórico de Germany
germany = df.loc[
    (df["Entity"] == "Germany") &
    (df["Annual change in population"].notna()),
    "Annual change in population"
]

# Estadísticas descriptivas
print(germany.describe())
print("Varianza:", germany.var())
```

### Países con cambio anual mínimo y máximo

Se identifican las entidades y los años correspondientes a los valores mínimo y máximo de `Annual change in population` de todo el DataFrame.

```python
columna_cambio = "Annual change in population"

# idxmin e idxmax devuelven los índices de los valores extremos, ignorando NaN
fila_min = df.loc[df[columna_cambio].idxmin(), ["Entity", "Year", columna_cambio]]
fila_max = df.loc[df[columna_cambio].idxmax(), ["Entity", "Year", columna_cambio]]

print("Cambio anual mínimo:")
print(fila_min)
print("\nCambio anual máximo:")
print(fila_max)
```
